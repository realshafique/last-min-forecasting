```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
individual_household_electric_power_consumption = fetch_ucirepo(id=235)

# data (as pandas dataframes)
X = individual_household_electric_power_consumption.data.features
y = individual_household_electric_power_consumption.data.targets

# metadata
print(individual_household_electric_power_consumption.metadata)

# variable information
print(individual_household_electric_power_consumption.variables)
```

```python
# ============================================================
# LAST MIN FORECASTING ⚡
# LSTM-Based Energy Consumption Forecasting
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint


# -----------------------------
# 2. LOAD DATA
# -----------------------------
# If X already exists from your UCI dataset-loading cell,
# this will use it.

data = X.copy()

print("Original shape:", data.shape)
print(data.head())


# -----------------------------
# 3. CREATE DATETIME INDEX
# -----------------------------

data["datetime"] = pd.to_datetime(
    data["Date"] + " " + data["Time"],
    format="%d/%m/%Y %H:%M:%S"
)

data.set_index("datetime", inplace=True)

data.drop(["Date", "Time"], axis=1, inplace=True)


# -----------------------------
# 4. CONVERT COLUMNS TO NUMERIC
# -----------------------------

numeric_cols = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

for col in numeric_cols:
    data[col] = pd.to_numeric(data[col], errors="coerce")


# -----------------------------
# 5. HANDLE MISSING VALUES
# -----------------------------

print("\nMissing values before cleaning:")
print(data.isnull().sum())

# Forward fill
data = data.ffill()

# Backward fill remaining values
data = data.bfill()

print("\nMissing values after cleaning:")
print(data.isnull().sum())


# -----------------------------
# 6. RESAMPLE TO HOURLY DATA
# -----------------------------

hourly_data = data.resample("h").mean()

# Remove any remaining missing rows
hourly_data.dropna(inplace=True)

print("\nHourly data shape:", hourly_data.shape)
print(hourly_data.head())


# -----------------------------
# 7. SELECT TARGET
# -----------------------------

target_column = "Global_active_power"

target_data = hourly_data[[target_column]].copy()

print("\nTarget statistics:")
print(target_data.describe())


# -----------------------------
# 8. TRAIN / VALIDATION / TEST SPLIT
# -----------------------------
# IMPORTANT:
# We split BEFORE fitting the scaler to prevent data leakage.

n = len(target_data)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

train_data = target_data.iloc[:train_end]
validation_data = target_data.iloc[train_end:validation_end]
test_data = target_data.iloc[validation_end:]

print("\nDataset split:")
print("Train:", train_data.shape)
print("Validation:", validation_data.shape)
print("Test:", test_data.shape)


# -----------------------------
# 9. SCALE DATA
# -----------------------------
# Fit scaler ONLY on training data.

scaler = MinMaxScaler(feature_range=(0, 1))

train_scaled = scaler.fit_transform(train_data)

validation_scaled = scaler.transform(validation_data)

test_scaled = scaler.transform(test_data)


# -----------------------------
# 10. CREATE SEQUENCES
# -----------------------------

TIME_STEP = 24


def create_sequences(data, time_step):
    X_seq = []
    y_seq = []

    for i in range(time_step, len(data)):
        X_seq.append(data[i-time_step:i, 0])
        y_seq.append(data[i, 0])

    return np.array(X_seq), np.array(y_seq)


X_train, y_train = create_sequences(
    train_scaled,
    TIME_STEP
)

X_validation, y_validation = create_sequences(
    validation_scaled,
    TIME_STEP
)

X_test, y_test = create_sequences(
    test_scaled,
    TIME_STEP
)


# -----------------------------
# 11. RESHAPE FOR LSTM
# -----------------------------

X_train = X_train.reshape(
    X_train.shape[0],
    X_train.shape[1],
    1
)

X_validation = X_validation.reshape(
    X_validation.shape[0],
    X_validation.shape[1],
    1
)

X_test = X_test.reshape(
    X_test.shape[0],
    X_test.shape[1],
    1
)


print("\nSequence shapes:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_validation:", X_validation.shape)
print("y_validation:", y_validation.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)


# -----------------------------
# 12. BUILD LSTM MODEL
# -----------------------------

model = Sequential([

    LSTM(
        64,
        return_sequences=True,
        input_shape=(TIME_STEP, 1)
    ),

    Dropout(0.2),

    LSTM(
        64,
        return_sequences=False
    ),

    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(1)
])


# -----------------------------
# 13. COMPILE MODEL
# -----------------------------

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

model.summary()


# -----------------------------
# 14. CALLBACKS
# -----------------------------

os.makedirs("models", exist_ok=True)

checkpoint_path = "models/best_lstm_model.keras"

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

checkpoint = ModelCheckpoint(
    checkpoint_path,
    monitor="val_loss",
    save_best_only=True
)


# -----------------------------
# 15. TRAIN MODEL
# -----------------------------

history = model.fit(

    X_train,
    y_train,

    validation_data=(
        X_validation,
        y_validation
    ),

    epochs=30,

    batch_size=64,

    shuffle=False,

    callbacks=[
        early_stopping,
        checkpoint
    ],

    verbose=1
)


# -----------------------------
# 16. LOAD BEST MODEL
# -----------------------------

from tensorflow.keras.models import load_model

model = load_model(checkpoint_path)

print("\nBest model loaded.")


# -----------------------------
# 17. PREDICT TEST DATA
# -----------------------------

predictions_scaled = model.predict(
    X_test
)


# -----------------------------
# 18. INVERSE TRANSFORM
# -----------------------------

predictions = scaler.inverse_transform(
    predictions_scaled
)

actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)


# -----------------------------
# 19. EVALUATION
# -----------------------------

mae = mean_absolute_error(
    actual,
    predictions
)

rmse = np.sqrt(
    mean_squared_error(
        actual,
        predictions
    )
)

print("\n==============================")
print("MODEL PERFORMANCE")
print("==============================")

print(f"MAE  : {mae:.4f} kW")
print(f"RMSE : {rmse:.4f} kW")


# -----------------------------
# 20. PLOT TRAINING HISTORY
# -----------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training History")

plt.xlabel("Epoch")

plt.ylabel("MSE Loss")

plt.legend()

plt.show()


# -----------------------------
# 21. ACTUAL VS PREDICTED
# -----------------------------

plt.figure(figsize=(15, 6))

plt.plot(
    actual[:300],
    label="Actual"
)

plt.plot(
    predictions[:300],
    label="Predicted"
)

plt.title(
    "Actual vs Predicted Energy Consumption"
)

plt.xlabel("Time")

plt.ylabel("Global Active Power (kW)")

plt.legend()

plt.show()


# -----------------------------
# 22. SAVE FINAL MODEL
# -----------------------------

final_model_path = "models/last_min_forecasting_lstm.keras"

model.save(
    final_model_path
)

print(
    f"\nModel saved to: {final_model_path}"
)


# -----------------------------
# 23. SAVE SCALER
# -----------------------------

scaler_path = "models/lstm_scaler.joblib"

joblib.dump(
    scaler,
    scaler_path
)

print(
    f"Scaler saved to: {scaler_path}"
)


# -----------------------------
# 24. SAVE MODEL CONFIGURATION
# -----------------------------

config = {
    "target_column": target_column,
    "time_step": TIME_STEP,
    "forecast_horizon": 1,
    "frequency": "hourly",
    "unit": "kW"
}

joblib.dump(
    config,
    "models/model_config.joblib"
)

print(
    "Model configuration saved."
)


# -----------------------------
# 25. FINAL FILE CHECK
# -----------------------------

print("\n==============================")
print("SAVED FILES")
print("==============================")

print(os.listdir("models"))
```
